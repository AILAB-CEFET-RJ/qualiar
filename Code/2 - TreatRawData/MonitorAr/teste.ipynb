```python
import pandas as pd
from io import BytesIO
import certifi
import requests
```

```python
url_internacoes = "https://raw.githubusercontent.com/AILAB-CEFET-RJ/qualiar/refs/heads/Refactoring-And-Documentation/Data/RawData/MonitorAr/dados_qualidade_ar_sensores.parquet"

response = requests.get(url_internacoes, verify=certifi.where(), timeout=60)
response.raise_for_status()

df_internacoes = pd.read_parquet(BytesIO(response.content))
```
